### Structure output


```python
import os

from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.getenv("Groq_Api_Key")
)
model
```

```python
from pydantic import BaseModel,Field

# output that LLM should be generating

class Movie(BaseModel):
    title:str=Field(description="the title of the movie")
    year:int=Field(description="this year the movie was released")
    director:str=Field(description="this is the person who directed the movie")
    rating:float=Field(description="the movies rating out of 10")
```

```python
model_with_structure=model.with_structured_output(Movie)
model_with_structure
```

### TypedDict

```python
from typing_extensions import TypedDict,Annotated

class Movie(TypedDict):
    title:Annotated[str,"the title of the movie"]
    year:Annotated[int,"the year the movie was released"]
    director:Annotated[str,"the director the movie"]
    rating:Annotated[float,"the rating of the movie out of 10"]


response=model.with_structured_output(Movie)
newResponse=response.invoke("please provide the details of the movie avengers")
newResponse
```
